## Data Exploration

This notebook performs exploratory data analysis (EDA) on the credit risk dataset. The goal is to understand the data structure, identify missing values, examine target distribution, and prepare for feature engineering.

```python
import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/statlog/german/german.data"

columns = [
    'Status_Checking', 'Duration', 'Credit_History', 'Purpose',
    'Credit_Amount', 'Savings', 'Employment_Since',
    'Installment_Rate', 'Personal_Status_Sex', 'Other_Debtors',
    'Residence_Since', 'Property', 'Age', 'Other_Installment_Plans',
    'Housing', 'Existing_Credits', 'Job', 'Dependents',
    'Telephone', 'Foreign_Worker', 'Target'
]

df = pd.read_csv(url, sep=' ', header=None, names=columns)

df.head()
```
